```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn


DATA_URL = (
    "https://raw.githubusercontent.com/coopersigrist/SunSight/"
    "99588d1696c980075a19abf087c669ae5aa22e4e/"
    "Data/Clean_Data/data_by_zip.csv"
)

SUN = "yearly_sunlight_kwh_kw_threshold_avg"
CARBON = "carbon_offset_metric_tons"
POP = "Total_Population"
INSTALLS = "existing_installs_count"
INCOME = "Median_income"
KEYS = [INCOME, INSTALLS, "black_prop", "asian_prop", SUN, CARBON, POP]
LABELS = ["Median income", "Installations", "Black proportion", "Asian proportion",
          "Solar yield", "Total carbon potential", "Population"]


def load_and_clean(path="data_by_zip.csv"):
    try:
        raw = pd.read_csv(path)
        print("Data source:", path)
    except FileNotFoundError:
        raw = pd.read_csv(DATA_URL)
        print("Local CSV not found. Using author repository snapshot:", DATA_URL)

    print("Original dimensions:", raw.shape)
    numeric = raw.select_dtypes(include=np.number)
    print("Missing values:", int(raw.isna().sum().sum()))
    print("Infinite numeric values:", int(np.isinf(numeric).sum().sum()))

    filled = raw.replace([np.inf, -np.inf], np.nan).fillna(0)
    numeric = filled.select_dtypes(include=np.number)
    keep = (numeric - numeric.mean()).abs().le(3 * numeric.std(ddof=1)).all(axis=1)
    clean = filled.loc[keep].reset_index(drop=True)
    if clean.empty:
        raise ValueError("No rows remain after cleaning.")
    clean.to_csv("clean_data_by_zip.csv", index=False)
    print("Cleaned dimensions:", clean.shape)
    print("Rows removed:", len(raw) - len(clean))
    return clean


def state_mask(frame, abbreviation, full_name):
    states = frame["state_name"].astype(str).str.strip().str.upper()
    return states.isin([abbreviation.upper(), full_name.upper()]).to_numpy()


def paired_values(x, y):
    x, y = np.asarray(x, dtype=float), np.asarray(y, dtype=float)
    if x.ndim != 1 or y.ndim != 1 or x.shape != y.shape:
        raise ValueError("Inputs must be one-dimensional arrays of equal length.")
    return x, y, np.isfinite(x) & np.isfinite(y)


def scatter_plot(x_values, y_values, stats_line=None, masks=None):
    x, y, valid = paired_values(x_values, y_values)
    if stats_line not in (None, "mean", "median"):
        raise ValueError("stats_line must be None, 'mean', or 'median'.")
    masks = [] if masks is None else masks
    fig, ax = plt.subplots(figsize=(8, 5), layout="constrained")
    covered = np.zeros(len(x), dtype=bool)
    for mask, name, color in masks:
        covered |= np.asarray(mask, dtype=bool)
    other = valid & ~covered
    ax.scatter(x[other], y[other], s=10, alpha=0.4,
               color="tab:blue", label="Other ZIP codes" if masks else None)
    for mask, name, color in masks:
        selected = valid & np.asarray(mask, dtype=bool)
        ax.scatter(x[selected], y[selected], s=10, alpha=0.6,
                   color=color, label=name)
    if stats_line is not None and valid.any():
        statistic = np.mean if stats_line == "mean" else np.median
        ax.axvline(statistic(x[valid]), color="gray", linestyle=":", alpha=0.5)
        ax.axhline(statistic(y[valid]), color="gray", linestyle=":", alpha=0.5)
    if masks:
        ax.legend()
    return ax


def complex_scatter_plot(x_values, y_values, stats_line=None, masks=None):
    masks = [] if masks is None else masks
    ax = scatter_plot(x_values, y_values, stats_line,
                      [(mask, name, color) for mask, name, color, degree in masks])
    x, y, valid = paired_values(x_values, y_values)
    for mask, name, color, degree in masks:
        if not isinstance(degree, (int, np.integer)) or degree < 0:
            raise ValueError("Polynomial degree must be a nonnegative integer.")
        selected = valid & np.asarray(mask, dtype=bool)
        xm, ym = x[selected], y[selected]
        if np.unique(xm).size <= degree:
            print(f"Skipping {name}: insufficient distinct x values for degree {degree}.")
            continue
        polynomial = np.polynomial.Polynomial.fit(xm, ym, degree)
        grid = np.linspace(xm.min(), xm.max(), 200)
        ax.plot(grid, polynomial(grid), color=color, alpha=0.5,
                label=f"{name}: degree {degree}")
    if masks:
        ax.legend()
    return ax


def pearson_corr(X, Y):
    x, y, valid = paired_values(X, Y)
    x, y = x[valid], y[valid]
    if len(x) < 2:
        return np.nan
    x, y = x - x.mean(), y - y.mean()
    denominator = np.sqrt(np.dot(x, x)) * np.sqrt(np.dot(y, y))
    if denominator == 0:
        return np.nan
    return float(np.clip(np.dot(x, y) / denominator, -1, 1))


def correlation_report(frame):
    rows = []
    for name, subset in {
        "All available ZIPs": frame,
        "CA": frame.loc[state_mask(frame, "CA", "California")],
        "MA": frame.loc[state_mask(frame, "MA", "Massachusetts")],
    }.items():
        population = subset[POP]
        per_capita = subset[INSTALLS] / population.where(population > 0)
        rows.append({
            "Scope": name,
            "Population vs installations": pearson_corr(population, subset[INSTALLS]),
            "Income vs installations per capita": pearson_corr(subset[INCOME], per_capita),
        })
    return pd.DataFrame(rows).set_index("Scope")


def pearson_matrix(dataframe, keys):
    matrix = np.full((len(keys), len(keys)), np.nan)
    for i, first in enumerate(keys):
        for j in range(i, len(keys)):
            matrix[i, j] = matrix[j, i] = pearson_corr(dataframe[first], dataframe[keys[j]])
    return pd.DataFrame(matrix, index=keys, columns=keys)


def draw_heatmap(matrix, title, limit=1):
    fig, ax = plt.subplots(figsize=(10, 8), layout="constrained")
    labels = LABELS if list(matrix.columns) == KEYS else list(matrix.columns)
    sn.heatmap(matrix, annot=True, fmt=".2f", cmap="coolwarm", center=0,
               vmin=-limit, vmax=limit, square=True, ax=ax,
               xticklabels=labels, yticklabels=labels)
    ax.set_title(title)
    ax.tick_params(axis="y", rotation=0)
    plt.setp(ax.get_xticklabels(), rotation=40, ha="right")
    return ax


def pearson_heatmap(dataframe, keys):
    correlations = pearson_matrix(dataframe, keys)
    shown = correlations.to_numpy(copy=True)
    pairs = shown[np.triu_indices(len(keys), k=1)]
    finite = pairs[np.isfinite(pairs)]
    np.fill_diagonal(shown, finite.mean() if finite.size else np.nan)
    draw_heatmap(pd.DataFrame(shown, index=keys, columns=keys),
                 "Pearson correlations; diagonal = mean of valid unique pairs")
    return correlations


def compare_state_heatmaps(frame):
    scopes = {"All available ZIPs": frame}
    for code, name in [("CA", "California"), ("MA", "Massachusetts"), ("MI", "Michigan")]:
        scopes[code] = frame.loc[state_mask(frame, code, name)]
    correlations = {}
    for name, subset in scopes.items():
        correlations[name] = pearson_heatmap(subset, KEYS)
        plt.gca().set_title(f"{name}: Pearson correlations\nDiagonal = mean of valid unique pairs")
        plt.show()

    differences = {}
    for state in ("CA", "MA", "MI"):
        difference = correlations[state] - correlations["All available ZIPs"]
        np.fill_diagonal(difference.values, 0)
        differences[state] = difference
    limit = max(0.01, max(np.nanmax(np.abs(d.values)) for d in differences.values()))
    upper = np.triu_indices(len(KEYS), k=1)
    for state, difference in differences.items():
        draw_heatmap(difference, f"{state} minus all available ZIPs", limit)
        plt.show()
        values = difference.to_numpy()[upper]
        finite = np.flatnonzero(np.isfinite(values))
        if not finite.size:
            print(f"{state}: no defined off-diagonal differences.")
            continue
        k = finite[np.argmax(np.abs(values[finite]))]
        i, j = upper[0][k], upper[1][k]
        print(f"{state}: largest absolute difference: {KEYS[i]} vs {KEYS[j]}; "
              f"state r={correlations[state].iloc[i, j]:+.4f}, "
              f"all-ZIP r={correlations['All available ZIPs'].iloc[i, j]:+.4f}, "
              f"difference={values[k]:+.4f}")
    return correlations, differences


def concluding_analysis(frame):
    resource_r = pearson_corr(frame[SUN], frame[INSTALLS])
    answers = {
        "8a": (
            f"Solar yield and installation counts have r={resource_r:+.4f} in this cleaned sample. "
            "Read this alongside the income/adoption correlations reported in Section 6. "
            "Positive values indicate associations with solar resource and ZIP-level income, "
            "not causal effects or proof of an efficient allocation. The requested carbon column "
            "is total annual technical carbon-abatement potential across suitable roof space, "
            "not emissions avoided per kWh or per panel. Its relationship with installations "
            "therefore mixes location quality with ZIP size and roof capacity. These plots "
            "cannot establish the existing fleet's carbon efficiency or Pareto optimality. "
            "Paper 1, Sections 3 and 6, provides separate evidence that carbon-aware siting "
            "can outperform existing deployment trends in simulations. Our sample is restricted "
            "to covered ZIPs surviving zero-imputation and all-column outlier filtering; it "
            "does not represent all US ZIP codes."
        ),
        "8b": (
            "Let P_i be number_of_panels_total, T_i be yearly_sunlight_kwh_total, and B_i "
            "be carbon_offset_metric_tons. Under a uniform-average-panel approximation, use "
            "E_i=T_i/P_i kWh per panel per year and C_i=B_i/P_i metric tons per panel per year "
            "for P_i>0. Do not multiply B_i by sunlight: B_i already includes generation. "
            "Choose positive reference scales E_ref and C_ref and an explicit policy weight "
            "0<alpha<1. Maximize sum_i a_i*(alpha*E_i/E_ref+(1-alpha)*C_i/C_ref), subject "
            "to integer allocations 0<=a_i<=u_i and sum_i a_i=n, where u_i is verified "
            "remaining panel capacity. Rank ZIPs by score and fill each to its capacity. "
            "For equal-cost, identical units with fixed marginal values and only these "
            "constraints, exchanging a unit from a lower-score ZIP to an available "
            "higher-score ZIP never decreases the objective, which proves greedy optimality. "
            "Different weights expose energy-carbon trade-offs; weighted sums can miss "
            "unsupported Pareto points in a discrete problem. Existing installs count rooftop "
            "systems/buildings, not individual panels, so subtracting existing_installs_count "
            "from number_of_panels_total does not give valid remaining panel capacity. "
            "Total roof capacity is only an optimistic upper bound without existing panel "
            "inventories. Average technical yields, fixed grid emission factors, equal costs, "
            "and available capacity are assumptions, not verified marginal deployment outcomes."
        ),
    }
    for section, answer in answers.items():
        print(f"\n{section}\n{answer}")
    print("\nSources:")
    print("Paper 1: https://doi.org/10.1145/3679240.3734603")
    print("Paper 2: https://doi.org/10.1145/3736425.3770115 (research extension, not implemented here)")
    print("Sunroof methodology: https://www.google.com/get/sunroof/data-explorer/")
    print("Dataset snapshot:", DATA_URL)
    return answers


def main(path="data_by_zip.csv"):
    df = load_and_clean(path)
    ca = [(state_mask(df, "CA", "California"), "CA ZIP codes", "tab:orange")]
    states = [(state_mask(df, "CA", "California"), "CA ZIP codes", "green"),
              (state_mask(df, "MA", "Massachusetts"), "MA ZIP codes", "black")]
    solar_labels = dict(xlabel="Annual solar yield (kWh/kW/year)",
                        ylabel="Total annual carbon potential (metric tons CO2e/year)",
                        title="Solar yield and total carbon-abatement potential")
    population_labels = dict(xlabel="ZIP population", ylabel="Existing rooftop installations",
                             title="Population and solar installations")

    for masks, stat in [(None, None), (ca, None), (ca, "mean")]:
        ax = scatter_plot(df[SUN], df[CARBON], stats_line=stat, masks=masks)
        ax.set(**solar_labels)
        plt.show()

    plots = [(SUN, CARBON, "mean", ca, solar_labels),
             (POP, INSTALLS, "median", states, population_labels)]
    for xkey, ykey, stat, masks, labels in plots:
        ax = scatter_plot(df[xkey], df[ykey], stats_line=stat, masks=masks)
        ax.set(**labels)
        plt.show()

    for xkey, ykey, stat, masks, labels in plots:
        fitted_masks = [(mask, name, color, 4) for mask, name, color in masks]
        ax = complex_scatter_plot(df[xkey], df[ykey], stats_line=stat, masks=fitted_masks)
        ax.set(**labels)
        plt.show()

    report = correlation_report(df)
    print("\nSection 6: Pearson correlations")
    print(report.round(4).to_string())
    correlations, differences = compare_state_heatmaps(df)
    answers = concluding_analysis(df)
    return df, report, correlations, differences, answers


if __name__ == "__main__":
    main()
```
